```python
import numpy as np

#2x^3-5x^2+3x+1
#buat fungsi nya
f = lambda x: 2*x**3 - 5*x**2 + 3*x + 1


#di soal minta trapezoid dan left reimann



#buat batas atas dan batas bawah
a = -2
b = 2


#grid point langsung pake, grid ditambah 1 dulu buat jadi n nya
n = 20 #titik , point, kalau diketahui itu grid, angka di soal + 1 = n
x = np.linspace(a,b,n) # n di sini itu artinya membagi berapa titik
y = f(x)
h = (b - a) / (n - 1) # n -1 itu artinya jumlah grid yang dipake adalah 1 kurangnya dari titik yang ada

#karena left itu dimulai dari kiri [:n-1]
ReimL = h * sum(y[:n-1])
print(ReimL)

#trapezoid itu pakai rumus luas trapesium
trap = h/2 * (y[0] + y[n-1] + sum(2 * y[1:n-1]))
print(trap)


#ada kemungkinan keluar error nya

#jawabannya itu nanti ada contoh output TAPI BEDA THRESHOLD, !!JANGAN LUPA DIGANTI KEMBALI jadi pakai thershold nim kalian!!
```
