# Newton Raphson

Menebak nilai x yang cocok untuk polinom, untuk hasilnya mendekati nol

```python
import numpy as np
```

##### mendapat nilai x baru = f(x)/f'(x)

```python
f = lambda x: x**5 + 6*x**4 + 10*x**3 - 21*x**2 - 40*x + 43
# f ini fungsi f(x)

g = lambda x: 5*x**4 + 24*x**3 + 30*x**2 - 42*x - 40
```

```python
#definisi variable
initial_value = -10
tolerance = 0.093
max_iter = 73
```

```python
def newton_raphson(x, iter):
    # update nilai x, dengan rumus newton rapshn
    x_new = x - f(x)/g(x)
    print(f"Iteration {iter}: {x_new}")
    
    #kalau sudah ketemu nilai f nya kurang dari toleransi, return
    if np.abs(f(x_new)) < tolerance:
        print(f"Root found at {x_new}")
        return x_new
    
    # validasi kalau lebih dari max iter
    if (iter >= max_iter):
        print("Max iter reached")
        return
    
    
    return newton_raphson(x_new, iter+1)
    
```
