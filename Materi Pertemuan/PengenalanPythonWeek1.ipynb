```python
# list
list1 = [1,2,3,4,5] #mutable (bisa diganti append atau pop [hapus data di index tertentu]), ordered
# ada index dari 0 

for i in list1:
    print(i)


list1.append(6) #di akhir
list1.insert(3, 10)

print("Setelah di append:")
for i in list1:
    print(i)

print("")

list1.pop(4)
print("Setelah di pop: ")
for i in list1:
    print(i)
 
```
