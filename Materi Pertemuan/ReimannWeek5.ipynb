```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
    return 6*(x**3) + 5*(x**2) - 7*x
```

```python
# buat batas awal dan batas akhir
b = 0 #start point
a = 3 #end point
n = 10
h = (a - b) / (n-1) # jarak Δx 

x = np.linspace(b,a,n) #ada 10 titik, berarti 9 kotak / bagian !! Tergantung soal mintanya titik atau kotak
y = f(x)

base_val = 150 #ini biasanya exact valuenya trus cari errornya


print(h)   

#reimann Left (dari awal sampe n -1, n nya kaga ke pake)
riemL = h * sum(y[:n-1])
errorL = base_val - riemL
print(riemL)
print(errorL)



#reimann Right (dari awal + 1 sampe akhir)
riemR = h * sum(y[1:])
errorR = base_val - riemR
print(riemR)
print(errorR)



#reimann Middle (cari mid point tiap persegi panjang dulu, cari f(xmidpoint baru di sum))
riemM = h * sum(f((x[:n-1] + x[1:])/2))
errorM = base_val - riemM
print(f'Luas pakai Reimann Middle: {riemM}')
print(errorM)


# kaga perlu di tampilin gak papa
plt.plot(x,y)
plt.show() # buat hilangin header text tidak jelas
```
