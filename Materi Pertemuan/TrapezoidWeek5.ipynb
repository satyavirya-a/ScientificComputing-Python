```python
import numpy as np

def f(x): 
    return 6*(x**3) + 5*(x**2) - 7*x

#berbeda dengan reimann yang pake persegi panjang, trapezoid itu pake trapesium
```

```python
b = 0
a = 3
n = 10 # n itu berarti berapa titik data x
h = (a - b)/(n-1)

x = np.linspace(b,a,n)
y = f(x)

trap = h/2 * (y[0] + y[n - 1] + sum(2 * y[1:n-1])) # nilai y awal dan terakhir itu cuma dipake sekali, sisanya dipake 2 kali makanya sumnya 2*y
print(trap)
```
