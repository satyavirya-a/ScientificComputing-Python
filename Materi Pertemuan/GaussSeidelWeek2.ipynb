```python
# buat fungsi cari diagonally dominant

def is_not_diagonally_dominant(A):
    diag = np.diag(np.abs(A))
    off_diagonal = np.sum(np.abs(A), axis = 1) - diag # axis = 1 itu untuk ambil x kesamping
                                                    # kalau axis = - 1 itu ambil x kebawah
    
    # cek semua baris pake fungsi ini
    if np.all(diag >= off_diagonal):
        return False # diagonal dominant (karena fungsinya NOT)
    else:
        return True # not diagonal dominant
    
```

```python
# buat fungsi gauss_seidel
def gauss_seidel(matrix_A, matrix_B, max_iters = 16, tolerance = 0.099):
    xs = np.array(matrix_A)
    ys = np.array(matrix_B) # ys matriks sisi kanan (setelah =)
    
    # cek dulu apakah matrix diagonally dominance
    if is_not_diagonally_dominant(xs):
        print("Matrix tidak diagonal dominan")
        return
    
    diag = np.diag(xs) # ambil semua value diagonal 
    xs = - xs # copy hal yang sama tadi tapi tandanya berubah
    np.fill_diagonal(xs, 0) # buat matrix diagonal = 0
    
    x_old = np.zeros(xs.shape[0]) # shape itu ukuran matrix nya, jadi dia ambil ukuran shape 3 dari barus 1nya
    
    # lakukan iterasi sampai iterasi yang dimau
    for idx in range(max_iters):
        # update nilai x_new
        x_new = np.array(x_old)
        
        for j, row in enumerate(xs): # kenapa pake enumerate (looping data sampai 2 value, idx dan valuenya)
            # hasil enumerate (idx, [row matriks] [row matrix])
            x_new[j] = (ys[j] + np.dot(row, x_new)) / diag[j]
            
        x_new = np.array(x_new)
            
        #print iterasi
        print(f"Iterasi ke-{idx}")
        print(f"\tx_new: {x_new}")
        
       
        error = np.sqrt((x_new - x_old)**2)/(np.sqrt((x_new**2)))
        # error = (np.abs(x_new) - np.abs(x_old))/np.abs(x_new)
        print(f"error relatif {x_new} {x_old} : {error}")
            
    
        
        # cek apakah konvergen atau divergen
        if(np.sqrt(np.sum((x_new - x_old)**2)) < tolerance):
            print("Konvergen") 
            return x_new
        
        x_old = x_new
        
    
```

```python
def main(): 
    for idx, (matrix_A, matrix_B) in enumerate(zip(x,y)):
        print(f"Persamaan ke-{idx+1}")
        # panggil fungsi cek diagonal dominance
        print(f"Matriks atau persamaan adalah {is_not_diagonally_dominant(np.array(matrix_A))}")
      
        gauss_seidel(matrix_A,matrix_B)
        print()


```
