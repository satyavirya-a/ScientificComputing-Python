```python
# Direct Inverse Method

x = [1,2,3,4,5,6,7,8]
y = [2,5,1,7,9,3,6,10]

# buat np.array dulu supaya make sure ini angka dan ubah ini jadi object yang bisa DIOPERASI METHOD di numpya
x = np.array(x)
y = np.array(y)

plt.plot(x,y, 'ro' )


# puter x nya
A = np.vstack([x, np.ones(len(x))]).T # susun vertikal antara matrix x dan matrix ones

# np.ones itu buat 1 baris isinya 1 semua, (len(x)) itu kolom baris 1 nya sepanjang x
# .T itu untuk transpose
#1. 2. itu karena dia flot
# print(A)

# print(y) # sebelum di tambahin axis
y = y[:, np.newaxis] # ':' untuk ambil 1 1 semuaBaris, ubah jadi vertikal dan 2D Array
# print(y) # sesudah ditambah axis

# np.dot(A.T, A)
# np.linalg.inv(np dot diatas)
# np.dot(inversnya, A.T)
# np.dot(hasil atas, y)

#jadinya matrix [m, b]
# alpha = np.dot(np.dot(np.linalg.inv(np.dot(A.T, A)), A.T), y)
# print(alpha)

#ada shortcut psudo invers di numpy
pinv = np.linalg.pinv(A)
# print(f"pinv {pinv}")
alpha = np.dot(pinv, y) # hasilnya matrix [m, b]
# print(f"alpha {alpha}")

plt.plot(x, alpha[0]*x + alpha[1])

plt.show()
```

```python
# Polynomial Regression

x = [1,2,3,4,5,6,7,8]
y = [2,5,1,7,9,3,6,10]

# buat np.array dulu supaya make sure ini angka dan ubah ini jadi object yang bisa DIOPERASI METHOD di numpya
x = np.array(x)
y = np.array(y)

# menggunakan titik y sebelumnya untuk mengestimasi nilai y selanjutnya
for i in range(1, 7): # 7 kali buat orde iterasinya
    
    y_est = np.polyfit(x,y,i) # mencari koefisien polinomial untuk nilai x agar nilai y nya sama titik point data
    
    # 2 baris
    # 3 kolom
    plt.subplot(2,3,i) # buat taruh grafiknya diposisi mana 
    plt.plot(x,y, 'ro' )
    plt.plot(x, np.polyval(y_est, x)) # untuk menentukan nilai spesific ynya
    plt.title(f"No. {i}")
    plt.tight_layout() # biar rapi
    
plt.show()
    
```

```python
#Interpolation

from scipy.interpolate import CubicSpline

x = [1,2,3,4,5,6,7,8]
y = [2,5,1,7,9,3,6,10]

# buat np.array dulu supaya make sure ini angka dan ubah ini jadi object yang bisa DIOPERASI METHOD di numpya
x = np.array(x)
y = np.array(y)
plt.plot(x, y, 'ro')


# f itu fungsi bawaan yang bisa disesuaikan dengan data yang ada
f = CubicSpline(x,y)

x_new = np.linspace(1,8, 10) # memecah angka 1 (low limit) sampai 8 (high limit) menjadi 100 segment
y_new = f(x_new)

print(x_new)

plt.plot(x_new, y_new)



```
