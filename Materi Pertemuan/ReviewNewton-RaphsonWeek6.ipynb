```python
import numpy as np
```

```python
# BUAT FUNGSINYA
# f(x)=3x5+5x4−12x3+8x2+6x−10
f = lambda x: 3*x**5 + 5*x**4 - 12*x**3 + 8*x**2 + 6*x - 10
g = lambda x: 15*x**4 + 20*x**3 - 36*x**2 + 16*x + 6

# bisa juga langsung def f(x): return <persamaannya>
```

```python
#BUAT VARIABEL NYA yang diperlukan
initial_val = 6
max_iters = 87
tolerance = 0.093 # dari NIM 2902727293
```

```python
#newton raphson pakai REKURSIF
def newton_raphson(x, iter):
    
    x_new = x - f(x)/g(x)
    
    print(x_new)
    
    if (np.abs(f(x_new)) < tolerance): #toleransi pakai absolute f(x) "The error measurement should be |f(x)|."
        print("convergent")
        return x_new
    
    if (iter >= max_iters):
        print("Melebihi batas iterasi")
        return
    
    return newton_raphson(x_new, iter+1)
            
```

```python
newton_raphson(initial_val, 1)
```
