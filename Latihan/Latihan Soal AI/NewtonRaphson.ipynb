```python
import numpy as np

def f(x):
    return 3*x**4 - 4*x**3 + 2*x - 5

def df(x):
    return 12*x**3 - 12*x**2 + 2


val = 15
tolerance = 0.093
max_iters2 = 25

def newton_raphson(initial_val, iter):
    
    x_new = initial_val - f(initial_val)/df(initial_val)
    
    print(f"Iteration #{iter}: new root = {x_new} with result {f(x_new)}")
    
    if (np.abs(f(x_new)) < tolerance):
        # print("Converged")
        return x_new
    
    if (iter >= max_iters2):
        print("Max Iter reached")
        return
    
    return newton_raphson(x_new, iter + 1)
    
    
print(f"the root of equation is: {newton_raphson(val, 1)}")
```
