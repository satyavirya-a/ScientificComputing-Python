```python
import numpy as np

def f(x):
    return 5*x**4 - 2*x**3 + 4*x**2 - x

a = -4
b = 4
n = 80
x = np.linspace(a, b, n)
y = f(x)
h = (b - a)/ (n-1)

ReimM = h * np.sum(f((x[:n-1] + x[1:])/2))
print(ReimM)

Trap = h/2 * (y[0] + y[n-1] + np.sum(2* y[1:n-1])) # jangan lupa 2 kali nya
print(Trap)
```
