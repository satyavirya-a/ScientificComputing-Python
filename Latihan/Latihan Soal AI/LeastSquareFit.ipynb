```python
import numpy as np
import matplotlib.pyplot as plt

x = [1.5, 2.0, 2.5, 3.0, 3.5, 4.0, 4.5, 5.0, 5.5, 6.0]
y = [45, 52, 58, 60, 65, 70, 75, 82, 85, 95]

x = np.array(x)
y = np.array(y)

A = np.vstack([x, np.ones(len(x))]).T
# print(A)

y = y[:, np.newaxis]

pinv = np.linalg.pinv(A)
alpha = np.dot(pinv, y)

print(f"m = {alpha[0]}") #jangan lupa di print
print(f"c = {alpha[1]}")
plt.plot(x, y, 'ro')
plt.plot(x, alpha[0]*x + alpha[1])
plt.legend(["Data Points", "Least Square"])
plt.xlabel("Land Size")
plt.ylabel("House Price")
plt.show()

```
