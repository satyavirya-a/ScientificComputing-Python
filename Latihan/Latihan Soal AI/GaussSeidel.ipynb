```python
import numpy as np

x = [
    [
        [15, 2, -1, 1],
        [-2, 12, 3, -2],
        [1, -2, 20, 4],
        [3, 1, -2, 10]
    ],
    [
        [2, 8, -1],
        [10, -2, 3],
        [-1, 3, 5]
    ],
    [
        [8, 2, -2],
        [1, -6, 2],
        [2, 1, 9]
    ],
    [
        [10, -1, 2, 0],
        [-1, 11, -1, 3],
        [2, -1, 10, -1],
        [0, 3, -1, 8]
    ],
    [
        [1, 5, 1],
        [6, 2, 0],
        [1, 1, 5]
    ]
]

y = [
    [25, 18, 40, 15],
    [15, 20, 10],
    [8, -4, 12],
    [6, 25, -11, 15],
    [10, 5, 8]
]

def not_diagonally_dominance(A):
    diag = np.diag(np.abs(A))
    off_diag = np.sum(np.abs(A), axis=1) - diag
    
    if np.all(diag > off_diag):
        return False
    else:
        return True
    
def gauss_seidel(matrix_A, matrix_B, max_iters=10, tolerance= 0.093):
    
    xs = np.array(matrix_A)
    ys = np.array(matrix_B)
    
    if (not_diagonally_dominance(xs)):
        print("Not diagonally Dominance")
        return
    
    diag = np.diag(xs)
    xs = -xs
    np.fill_diagonal(xs, 0)
    
    x_old = np.zeros(xs.shape[0])
    
    for iter in range(max_iters):
        
        x_new = np.array(x_old)
        
        for j, row in enumerate(xs):
            x_new[j] = (ys[j] + np.dot(row, x_new)) / diag[j] #ini plus karena sebelumnya udah di min
            
        print(f"iter #{iter+1}: {x_new}")
        
        error = np.sqrt(np.sum((x_new - x_old)**2))
        
        if (error < tolerance):
            print("Converged")
            return x_new
        
        x_old = x_new
        
    print("Not Converged")
        
def main():
    for idx, (matrix_A, matrix_B) in enumerate(zip(x,y)):
        
        print(f"Equation #{idx + 1}:")
        gauss_seidel(matrix_A, matrix_B)
        print()

main()
        
        
        
```
