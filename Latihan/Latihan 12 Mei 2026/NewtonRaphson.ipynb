```python
import numpy as np
```

```python
f= lambda x: 3*x**5 + 5*x**4 - 12*x**3 + 8*x**2 + 6*x - 10
df= lambda x: 15*x**4 + 20*x**3 - 36*x**2 + 16*x + 6
```

```python
max_iter = 87
initial_val = 6
tol = 0.093
```

```python
def newton_raphson(x, iter):
    
    x_new = x - (f(x)/df(x))
    
    print(f"Iteration {iter}: {x_new} . With value of: {f(x_new)}")
    
    if (np.abs(f(x_new)) < tol):
        print("Found")
        return
    
    if (iter >= max_iter):
        print("Over the max iteration")
        return
    newton_raphson(x_new, iter + 1)
    
```

```python
newton_raphson(initial_val, 1)
```
