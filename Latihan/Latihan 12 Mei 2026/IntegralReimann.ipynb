```python
import numpy as np
```

```python
f = lambda x: 2*x**2 + x + 5
```

```python
a = -7
b = 3
n = 30 # ini grid point

h = (b-a)/ (n-1) # di kurang 1 karena grid itu luasannya

x = np.linspace(a, b, n)
y = f(x)

RiemL = h * np.sum(y[:n-1])
print(f"Left Riemann    : {RiemL}")

RiemR = h * np.sum(y[1:])
print(f"Right Riemann    : {RiemR}")

RiemM = h * np.sum(f((x[:n-1] + x[1:])/2))
print(f"Midpoint Riemann    : {RiemM}")

Trap = h/2 * np.sum(y[0] + y[n-1] + 2*np.sum(y[1:n-1]))
print(f"Trapezoidal Rule    : {Trap}")

```
