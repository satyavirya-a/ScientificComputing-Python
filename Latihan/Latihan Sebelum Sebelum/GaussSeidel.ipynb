```python
x = [ 
    [ 
        [-61, 11, 7], 
        [20, 67, 8], 
        [38, 16, -98] 
    ], 
    [ 
        [-109, 14, 57], 
        [2, -64, 24], 
        [38, 4, -85] 
    ], 
    [ 
        [-62, 3, 13], 
        [17, -93, 39], 
        [20, 8, -69] 
    ], 
    [ 
        [105, 28, 112], 
        [56, 97, 84], 
        [20, 81, -60] 
    ], 
    [ 
        [-87, 23, 23], 
        [11, 95, 45], 
        [18, 44, -108] 
    ] 
] 
 
y = [ 
    [-969, 79, -265], 
    [78, -1, -382], 
    [-970, -849, -449], 
    [198, -314, 193], 
    [882, -17, 945] 
] 
```

```python
import numpy as np

#buat fungsi cek diagonal dominance dulu

def isNotDiagonallyDominant(A):
        #ubah angka jadi positif semua (Absolut) (di diag domi hanya nilai abs yang di cek)
        #ambil nilai diag nya
    diag = np.diag(np.abs(A))
    print("Matrix diag:")
    print(diag)
    
    print()
    
    #buat var matrix selain diag
        #jumlahkan nilai abs(A) secara horizontal (axis = 1)
        # kurangi dengan diag
    off_diag = np.sum(np.abs(A), axis = 1) - diag
    print("Matrix total offDiag:")
    print(off_diag)
    
    #cek semua element matrixnya, kalau semua diag >= off_diag berarti dia Diag Domi
    if np.all(diag >= off_diag):
        return False #dia diaogally dominace
    else:
        return True #dia TIDAK diagonaly dominance
    
    
    
def gauss_seidel(matrix_A, matrix_B, max_iters = 16, tolerance = 0.099):
    #buat jadi matrix dulu
    xs = np.array(matrix_A)
    ys = np.array(matrix_B)
    
    #cek apakah xs diagonal dominance 
    if (isNotDiagonallyDominant(xs)):
        print("Kaga diagonal Dominance")
        return
    
    #manipulasi koefisien seperti isolasi x di mtk
        #simpan nilai diag untuk pembagian nanti
        #ubah koef jadi negatif (-xs)
        #isi NOL di Koef diagonal
    diag = np.diag(xs)
    xs = -xs
    np.fill_diagonal(xs, 0)
    
    #BUAT TEBAKAN AWAL (Matrix 0 dengan shape 3*3 [xs.shape] baris 1 [0])
    x_old = np.zeros(xs.shape[0])
    
    #lakukan iterasi sampai yang dimau
    for idx  in range(max_iters):
        
        #update nilai x
        x_new = np.array(x_old)
        
        #iterasi solusi gauss seidel
        for j, row in enumerate(xs): #j itu idx dari enumerate, row itu baris dari matrix xs
                #buat x_new dari (b + (= x2*a2 - x3*a3) )/ diag LAKUKAN UNTUK TIAP BARIS
            x_new[j] = (ys[j] + np.dot(row, x_new)) / diag[j]
            
        print(f"Solusi x iterasi ke {idx}")
        print(x_new)
        
        #hitung error untuk toleransi
            #kuadratkan selisih x_new dan x_old
            #jumlah semua selisih tadi
            #akarkan jumlah tersebut
        error = np.sqrt(np.sum((x_new - x_old)**2))
        
        if (error < tolerance):
            print("Sudah Konvergen")
            return
        
        
        x_old = np.array(x_new) # jangan lupa update x_old setiap iterasi
        
            
    
def main():
    for idx, (matrix_A, matrix_B) in enumerate(zip(x,y)): #x dan y itu ada 5 SPL di soal, makanya di looping
            #urutan angka idx dari enumerate
            #zip(x,y) nyatuin elemen x dan y
            #matrix x itu matrix_A
            #matrix y itu matrix_B
        print(f"matrixA (Koefisien):\n {np.array(matrix_A)}")
        print(f"matrixB (hasil):\n {np.array(matrix_B)}")
        
        gauss_seidel(matrix_A, matrix_B)
        
        
        print("\n")
        
```

```python
main()
```
