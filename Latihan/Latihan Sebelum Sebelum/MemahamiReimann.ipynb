```python
import numpy as np

def f(x):
    return 2*x**3 - 5*x**2 + 3*x + 1

a = -2
b = 2
n = 20 #grid point berarti titik

x = np.linspace(a,b, n) # udah bener n karena garis dari a - b dipotong 20 titik
y = f(x)

h = (b - a) / (n - 1) # n - 1 karena jarak interval (h) itu selisih / jumlah grid

RiemL = h* np.sum(y[:n-1])
print(RiemL)

RiemR = h* np.sum(y[1:])
print(RiemR)

RiemM = h *  np.sum((f((x[1:] + x[:n-1])/2)))
print(RiemM)

Trap = h/2 * (y[0] + y[n-1] + np.sum(2* y[1:n-1]))
print(Trap)
```
