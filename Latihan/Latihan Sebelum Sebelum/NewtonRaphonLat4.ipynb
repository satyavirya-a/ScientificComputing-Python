```python
import numpy as np

def f(x):
    return 3*x**5 + 5*x**4 - 12*x**3 + 8*x**2 + 6*x - 10

def g(x):
    return 15*x**4 + 20*x**3 - 36*x**2 + 16*x + 6

initial_val = 6
max_iters = 87
tol = 0.093

def newton_raphson(initial_val, iter):
    x_new = initial_val - f(initial_val)/g(initial_val)
    
    print(x_new)
    
    if (np.abs(f(x_new)) < tol):
        return x_new
    
    if (iter > max_iters):
        print("Sudah melebihi max iters")
        return
    
    newton_raphson(x_new, iter + 1)
    

newton_raphson(initial_val, 1)
    
```
