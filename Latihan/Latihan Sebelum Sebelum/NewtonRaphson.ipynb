```python
import numpy as np

f = lambda x: 0.5*x**8 - 3*x**7 + 7*x **6 - 2*x **5 + 4*x **4 - x**3 + 3*x**2 - 6*x - 10
df = lambda x: 4*x ** 7 - 21*x ** 6 + 42*x ** 5 - 10*x ** 4 + 16*x ** 3 - 3*x**2 + 6*x - 6

initial_val = 6
max_iter = 10
tol = 0.0093
```

```python
def newton_raphson(x, iter):
    x_new = x - (f(x)/df(x))
    
    if (np.abs(f(x_new)) < tol):
        return x_new
    
    if (iter >= max_iter):
        return
    
    newton_raphson(x_new, iter + 1)
    
newton_raphson(initial_val, 1)
```
