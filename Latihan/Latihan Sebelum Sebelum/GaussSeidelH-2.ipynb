```python
def not_diagonally_dominance(A):
    diag = np.diag(np.abs(A))
    off_diag = np.sum(np.abs(A), axis=1) - diag
                            #axis 1 itu ambil yang di satu row yang sama
                            #axis 0 itu ambil satu col
                        
    
    if np.all(diag > off_diag):
        return False
        print("Tidak diagonal Dominance")
    else:
        print("Diagonal Dominance")
        return True
```

```python
def gauss_seidel(x,y, max_iters = 53, tol = 0.035):
    
    xs = np.array(x)
    ys = np.array(y)
    
    if (not_diagonally_dominance(xs)):
        return
    
    diag = np.diag(xs)
    x_old = np.zeros(xs.shape[0])
    
    xs = -xs
    np.fill_diagonal(xs, 0)
    
    for i in range(max_iters):
        
        x_new = np.array(x_old)
        
            #J buat index
        for j, row in enumerate(xs):
            x_new[j] = (ys[j] + np.dot(row, x_new)) / diag[j]
            #j = 1, index ke 1
            # row itu koefisien per baris
            
        print(x_new)
        error = np.sqrt(np.sum((x_new-x_old)**2))
        # print(error)
        if (error < tol):
            print("Konvergen")
            print(f"Penyelesaian x = {x_new} dengan error = {error}") #5 angka belakang koma
            return x_new
        
        x_old = x_new
```
