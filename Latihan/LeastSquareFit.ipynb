```python
import numpy as np
import matplotlib.pyplot as plt

x = [1,2,3,4,5,6,7,8]
y = [2,5,1,7,9,3,6,10]

x = np.array(x)
y = np.array(y)

#matrix yang nanti akan dijadikan bahan inverse
A = np.vstack([x, np.ones(len(x))]).T
print(A)

#buat biar bisa dikaliin nanti
y = y[:, np.newaxis]
print(y)

pinv = np.linalg.pinv(A) #A^-1 (ini inversnya A)
alpha = np.dot(pinv, y) # A*alpha = y -> alpha = A^-1 * y !!!

plt.plot(x, alpha[0]*x + alpha[1])

plt.plot(x,y, "ro")
plt.show()
```
