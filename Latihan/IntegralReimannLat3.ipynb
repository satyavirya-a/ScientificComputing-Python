```python
import numpy as  np

f = lambda x: 2*x**3 - 5*x**2 + 3*x + 1

n = 20
a = -2
b = 2

h = (b-a)/(n-1) 
# h ini jarak intervalnya
# n - 1 ini adalah jumlah grid atau  banyak intervalnya

x = np.linspace(a,b,n) #n karena kia pengen buat garis dari a sampe b dengan total titik pembagi adalah 20
y = f(x)

RiemL = h * np.sum(y[:n-1])
print(RiemL)

RiemR = h * np.sum(y[1:])
print(RiemR)

RiemM = h * np.sum(f((x[:n-1] + x[1:]) / 2))
print(RiemM)

Trapezoid = h/2* (y[0] + y[n-1] + np.sum(2*y[1:n-1]))
print(Trapezoid)
```
