```python
def not_diagonally_dominance(A):
    diag = np.diag(np.abs(A))
    off_diag = np.sum(np.abs(A), axis=1) - diag
    
    if np.all(diag > off_diag):
        return False
    else:
        return True
```

```python
def gauss_seidel(Matrix_A, Matrix_B, max_iter= 53, tol = 0.093):
    
    xs = np.array(Matrix_A)
    ys = np.array(Matrix_B)
    
    if (not_diagonally_dominance(xs)):
        return
    
    x_old = np.zeros(x.shape[0])
    
    diag = np.diag(xs)
    xs = -xs
    np.fill_diagonal(xs, 0)
    
    for iter in range(max_iter):
        
        x_new = np.array(x_old)
        
        for j, row in enumerate(xs):
            x_new[j] = (ys[j] + np.dot(row, x_new)) / diag[j]
            
        print(x_new)
        
        error = np.sqrt(np.dot(x_new - x_old, x_new - x_old))
        
        if (error < tol):
            return
        
        x_old = x_new
```
