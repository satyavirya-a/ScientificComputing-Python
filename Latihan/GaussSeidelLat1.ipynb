```python


x = [
    [
        [95, 12, -8],
        [7, -88, 14],
        [5, 9, -102]
    ],
    [
        [120, -15, 10],
        [8, -110, 6],
        [12, 7, -98]
    ],
    [
        [-105, 18, 9],
        [6, -92, 11],
        [10, 5, -115]
    ],
    [
        [85, -6, 4],
        [3, 97, 8],
        [7, 2, 90]
    ],
    [
        [-130, 25, 12],
        [14, -140, 20],
        [18, 10, -150]
    ]
]

y = [
    [210, -340, 415],
    [-520, 630, -710],
    [305, -450, 520],
    [150, 275, 390],
    [-610, 720, -830]
]
import numpy as np

def is_not_diagonally_dominance(A):
    diag = np.diag(np.abs(A))
    off_diag = np.sum(np.abs(A), axis=1) - diag
    
    if np.all(diag > off_diag):
        return False
    else:
        return True
    
    
def gauss_seidel(matrix_A, matrix_B, max_iters = 53, tolerance = 0.093):
    
    xs = np.array(matrix_A)
    ys = np.array(matrix_B)
    
    if (is_not_diagonally_dominance(xs)):
        print("Not diagonally Dominant")
        return
    
    diag = np.diag(xs)
    xs = -xs
    np.fill_diagonal(xs, 0)
    
    x_old = np.zeros(xs.shape[0])
    
    for iter in range(max_iters):
        
        x_new = np.array(x_old)
        
        for j, row in enumerate(xs):
            x_new[j] = (ys[j] - np.dot(row, x_new)) / diag[j]
            
        print(x_new)
        
        error = np.sqrt(np.sum((x_new - x_old)**2))
        
        if error < tolerance:
            print("Convergent")
            return
        
        x_old = x_new
        
    
def main():
    for idx, (matrix_A, matrix_B) in enumerate(zip(x,y)):
        
        gauss_seidel(matrix_A, matrix_B)
        
main()
        
        
```
