```python
import numpy as np

f = lambda x: 3*x**5 + 5*x**4 - 12*x**3 + 8*x**2 + 6*x - 10
g = lambda x: 15*x**4 + 20*x**3 - 36*x**2 + 16*x + 6

initial_val = 6
max_iteration = 87
tolerance = 0.093

def newton_raphson(val, iter):
    
    new_val = val - f(val)/g(val)
    print(f"Iteration-{iter}: {new_val}")
    
    if (iter > max_iteration):
        print("Melebihi batas iterasi")
        return
    
    if (np.abs(f(new_val)) < tolerance):
        print(f"Total Iteration: {iter}")
        print(f"Root of equation: {new_val}")
        return new_val
    
    newton_raphson(new_val, iter + 1)
    
newton_raphson(initial_val, 1)
    
```
