```python
import numpy as np

x = [ 
    [ 
        [-61, 11, 7], 
        [20, 67, 8], 
        [38, 16, -98] 
    ], 
    [ 
        [-109, 14, 57], 
        [2, -64, 24], 
        [38, 4, -85] 
    ], 
    [ 
        [-62, 3, 13], 
        [17, -93, 39], 
        [20, 8, -69] 
    ], 
    [ 
        [105, 28, 112], 
        [56, 97, 84], 
        [20, 81, -60] 
    ], 
    [ 
        [-87, 23, 23], 
        [11, 95, 45], 
        [18, 44, -108] 
    ] 
] 
 
y = [ 
    [-969, 79, -265], 
    [78, -1, -382], 
    [-970, -849, -449], 
    [198, -314, 193], 
    [882, -17, 945] 
] 
```

```python
def is_not_diagonally_dominance(A):
    diag = np.diag(np.abs(A))
    off_diag = np.sum(np.abs(A), axis=1) - diag
    
    if np.all(diag >= off_diag):
        return False
    else:
        return True

def gauss_seidel(matrix_A, matrix_B, max_iters=16, tolerance=0.099):
    xs = np.array(matrix_A)
    ys = np.array(matrix_B)
    
    if is_not_diagonally_dominance(xs):
        print("Tidak diagonally dominance")
        return
    
    diag = np.diag(xs)
    xs = -xs
    np.fill_diagonal(xs, 0)
    
    x_old = np.zeros(xs.shape[0])
    
    for iter in range(max_iters):
        
        x_new = np.array(x_old)
        
        for j, row in enumerate(xs):
            x_new[j] = (ys[j] + np.dot(row, x_new))/ diag[j]
            
        print(f"Solusi iterasi ke-{iter}")
        print(x_new)
        
        error = np.sqrt(np.sum((x_new - x_old)**2))
        
        if error < tolerance:
            print("konvergen")
            return
        
        x_old = np.array(x_new)
        
        
def main():
    
    for idx, (matrix_A,matrix_B) in enumerate(zip(x,y)):
        
        print(f"soal ke-{idx}")
        print(matrix_A)
        print(matrix_B)
        gauss_seidel(matrix_A, matrix_B)
        
        
main()  
```
