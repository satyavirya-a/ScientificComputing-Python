```python
import numpy as np
```

```python
def f(x):
    return 2*x**2 + x + 5
```

```python
a = -7
b = 3
n = 30

h = (b - a) / (n - 1) # di kurang 1 n nya karena jumlah grid = jumlah point - 1
x = np.linspace(a,b,n) # gak dikurang 1 itu karena garis nya dibagi berdasarkan point (titiknya)
y = f(x)

RiemL = h * np.sum(y[:n-1])
print(f"Left Riemann        : {RiemL}")

RiemR = h * np.sum(y[1:])
print(f"Right Riemann       : {RiemR}")

RiemM = h * np.sum(f((x[:n-1] + x[1:])/2))
print(f"Midpoint Riemann    : {RiemM}")

Trap = h/2 * (y[0] + y[n-1] + np.sum(2 * y[1:n-1]))
print(f"Trapezoidal Rule    : {Trap}")
```
