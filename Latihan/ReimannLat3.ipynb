```python
import numpy as np


def f(x):
    return 2*x**3 - 5*x**2 + 3*x + 1

a = -2
b = 2
n = 20 # ini grid point
h = (b-a)/(n-1)

x = np.linspace(a, b, n)
y = f(x)

RiemL = h * np.sum(y[:n-1])
print(RiemL)

RiemR = h * np.sum(y[1:])
print(RiemR)

RiemM = h* np.sum(f((x[1:] + x[:n-1])/2)) 
print(RiemM)

trap = h/2 * (y[0] + y[n-1] + np.sum(2 * y[1:n-1]))
print(trap)
```
